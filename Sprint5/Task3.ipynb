```

public delegate void WorkCompletedHandler(string taskName);
public delegate void VacationRequestedHandler(Employee employee, DateTime startDate, int days);

public class Employee
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Salary { get; set; }
    
    // Добавленные атрибуты
    public string Department { get; set; }
    public string Email { get; set; }
    public DateTime HireDate { get; set; }
    public int VacationDays { get; set; }
    
    // События
    public event WorkCompletedHandler WorkCompleted;
    public event VacationRequestedHandler VacationRequested;

    public Employee(string name, int age, int salary, string department, string email, DateTime hireDate)
    {
        Name = name;
        Age = age;
        Salary = salary;
        Department = department;
        Email = email;
        HireDate = hireDate;
        VacationDays = 28;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary} рублей, Отдел: {Department}";
    }

    public virtual void Work()
    {
        Console.WriteLine($"Сотрудник {Name} работает");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"Сотрудник {Name} взял отпуск");
        VacationDays -= 14;
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }
    
    // Добавленные методы
    public virtual void CompleteTask(string taskName)
    {
        Console.WriteLine($"Сотрудник {Name} завершил задачу: {taskName}");
        OnWorkCompleted(taskName);
    }
    
    public virtual void RequestVacation(DateTime startDate, int days)
    {
        if (days <= VacationDays)
        {
            Console.WriteLine($"Сотрудник {Name} запрашивает отпуск на {days} дней");
            OnVacationRequested(startDate, days);
        }
        else
        {
            Console.WriteLine($"Недостаточно дней отпуска. Доступно: {VacationDays}, запрошено: {days}");
        }
    }
    
    public int CalculateYearsInCompany()
    {
        return DateTime.Now.Year - HireDate.Year;
    }
    
    public virtual void AttendMeeting(string meetingTopic)
    {
        Console.WriteLine($"Сотрудник {Name} участвует в совещании: {meetingTopic}");
    }
    
    // Защищенные методы для вызова событий
    protected virtual void OnWorkCompleted(string taskName)
    {
        WorkCompleted?.Invoke(taskName);
    }
    
    protected virtual void OnVacationRequested(DateTime startDate, int days)
    {
        VacationRequested?.Invoke(this, startDate, days);
    }
}

public class Manager : Employee
{
    public int SubordinatesCount { get; set; }
    
    // Добавленные атрибуты
    public List<Employee> Team { get; set; }
    public string ManagementLevel { get; set; }
    public double BonusPercentage { get; set; }

    public Manager(string name, int age, int salary, int subordinatesCount, 
                  string department, string email, DateTime hireDate, 
                  string managementLevel, double bonusPercentage) 
                  : base(name, age, salary, department, email, hireDate)
    {
        SubordinatesCount = subordinatesCount;
        ManagementLevel = managementLevel;
        BonusPercentage = bonusPercentage;
        Team = new List<Employee>();
    }

    public override void Work()
    {
        Console.WriteLine($"Менеджер {Name} управляет командой из {SubordinatesCount} человек");
    }
    
    public override void Display()
    {
        Console.WriteLine($"{GetInfo()}, Количество подчиненных: {SubordinatesCount}, " +
                         $"Уровень управления: {ManagementLevel}, Бонус: {BonusPercentage}%");
    }
    
    // Добавленные методы
    public void AddToTeam(Employee employee)
    {
        Team.Add(employee);
        SubordinatesCount = Team.Count;
        Console.WriteLine($"Сотрудник {employee.Name} добавлен в команду менеджера {Name}");
    }
    
    public void ConductTeamMeeting()
    {
        Console.WriteLine($"Менеджер {Name} проводит собрание команды");
        foreach (var employee in Team)
        {
            employee.AttendMeeting($"Собрание команды менеджера {Name}");
        }
    }
    
    public double CalculateBonus()
    {
        return Salary * BonusPercentage / 100;
    }
    
    public void ReviewPerformance()
    {
        Console.WriteLine($"Менеджер {Name} проводит оценку эффективности команды");
        CompleteTask("Оценка эффективности команды"); 
    }
}

public class Developer : Employee
{
    public string Specialization { get; set; }
    
    // Добавленные атрибуты
    public string ProgrammingLanguage { get; set; }
    public int CompletedProjects { get; set; }
    public string GitHubProfile { get; set; }

    public Developer(string name, int age, int salary, string specialization,
                    string department, string email, DateTime hireDate,
                    string programmingLanguage, string githubProfile) 
                    : base(name, age, salary, department, email, hireDate)
    {
        Specialization = specialization;
        ProgrammingLanguage = programmingLanguage;
        GitHubProfile = githubProfile;
        CompletedProjects = 0;
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Разработчик {Name} взял отпуск, необходимо согласовать.");
    }
    
    public override void Display()
    {
        Console.WriteLine($"{GetInfo()}, Специализация: {Specialization}, " +
                         $"Язык программирования: {ProgrammingLanguage}, " +
                         $"Завершенные проекты: {CompletedProjects}");
    }
    
    // Добавленные методы
    public void WriteCode(string feature)
    {
        Console.WriteLine($"Разработчик {Name} пишет код для: {feature}");
    }
    
    public void CompleteProject(string projectName)
    {
        CompletedProjects++;
        Console.WriteLine($"Разработчик {Name} завершил проект: {projectName}");
        CompleteTask(projectName);
    }
    
    public void DebugCode()
    {
        Console.WriteLine($"Разработчик {Name} отлаживает код");
    }
    
    public override void AttendMeeting(string meetingTopic)
    {
        Console.WriteLine($"Разработчик {Name} участвует в техническом совещании: {meetingTopic}");
    }
}

public class Designer : Employee
{
    public string DesignStyle { get; set; }
    
    // Добавленные атрибуты
    public string DesignSoftware { get; set; }
    public int DesignProjectsCompleted { get; set; }
    public string PortfolioLink { get; set; }

    public Designer(string name, int age, int salary, string designStyle,
                   string department, string email, DateTime hireDate,
                   string designSoftware, string portfolioLink) 
                   : base(name, age, salary, department, email, hireDate)
    {
        DesignStyle = designStyle;
        DesignSoftware = designSoftware;
        PortfolioLink = portfolioLink;
        DesignProjectsCompleted = 0;
    }

    public override void Work()
    {
        Console.WriteLine($"Дизайнер {Name} работает над дизайн-проектом в стиле {DesignStyle}");
    }
    
    public override void Display()
    {
        Console.WriteLine($"{GetInfo()}, Стиль дизайна: {DesignStyle}, " +
                         $"Программное обеспечение: {DesignSoftware}, " +
                         $"Завершенные проекты: {DesignProjectsCompleted}");
    }
    
    // Добавленные методы
    public void CreateDesign(string projectName)
    {
        Console.WriteLine($"Дизайнер {Name} создает дизайн для: {projectName}");
    }
    
    public void CompleteDesignProject(string projectName)
    {
        DesignProjectsCompleted++;
        Console.WriteLine($"Дизайнер {Name} завершил дизайн-проект: {projectName}");
        CompleteTask(projectName); 
    }
    
    public void PresentDesign()
    {
        Console.WriteLine($"Дизайнер {Name} представляет дизайн клиенту");
    }
}

// Класс для управления сотрудниками
public class EmployeeManagementSystem
{
    private List<Employee> _employees;
    
    public EmployeeManagementSystem()
    {
        _employees = new List<Employee>();
    }
    
    public void AddEmployee(Employee employee)
    {
        _employees.Add(employee);
        
        employee.WorkCompleted += OnWorkCompleted;
        employee.VacationRequested += OnVacationRequested;
    }
    
    public void DisplayAllEmployees()
    {
        Console.WriteLine("\nВСЕ СОТРУДНИКИ");
        foreach (var employee in _employees)
        {
            employee.Display();
        }
    }
    
    public void ProcessWorkDay()
    {
        Console.WriteLine("\nРАБОЧИЙ ДЕНЬ");
        foreach (var employee in _employees)
        {
            employee.Work();
        }
    }
    
    private void OnWorkCompleted(string taskName)
    {
        Console.WriteLine($"[СИСТЕМА] Задача '{taskName}' завершена и зарегистрирована в системе");
    }
    
    private void OnVacationRequested(Employee employee, DateTime startDate, int days)
    {
        Console.WriteLine($"[СИСТЕМА] Запрос на отпуск от {employee.Name}: {days} дней с {startDate:dd.MM.yyyy}");
    }
    
    public void DisplayEmployeesByDepartment(string department)
    {
        var departmentEmployees = _employees.Where(e => e.Department == department);
        
        Console.WriteLine($"\nСОТРУДНИКИ ОТДЕЛА {department}");
        foreach (var employee in departmentEmployees)
        {
            employee.Display();
        }
    }
    
    public void DisplayHighSalaryEmployees(int minSalary)
    {
        var highSalaryEmployees = _employees.Where(e => e.Salary >= minSalary)
                                           .OrderByDescending(e => e.Salary);
        
        Console.WriteLine($"\nСОТРУДНИКИ С ЗАРПЛАТОЙ ОТ {minSalary}");
        foreach (var employee in highSalaryEmployees)
        {
            employee.Display();
        }
    }
}

// Создание системы управления сотрудниками
        var empSystem = new EmployeeManagementSystem();
        
        Manager manager = new Manager("Элиз Бентли", 39, 118000, 0, 
                                    "IT", "elise.bentley@company.com", new DateTime(2015, 3, 15),
                                    "Senior", 15.0);
        
        Developer developer = new Developer("Томас Мюллер", 37, 95000, "Веб-разработка",
                                          "IT", "thomas.muller@company.com", new DateTime(2018, 7, 22),
                                          "C#", "github.com/thomasmuller");
        
        Designer designer = new Designer("Кэтти Дейва", 29, 89000, "Модерн",
                                       "Design", "kathy.dayva@company.com", new DateTime(2020, 1, 10),
                                       "Figma", "behance.net/kathydayva");
        
        empSystem.AddEmployee(manager);
        empSystem.AddEmployee(developer);
        empSystem.AddEmployee(designer);
        
        manager.AddToTeam(developer);
        
        empSystem.DisplayAllEmployees();
        empSystem.DisplayEmployeesByDepartment("IT");
        empSystem.DisplayHighSalaryEmployees(90000);
        
        empSystem.ProcessWorkDay();
        
        Console.WriteLine("\nВЫПОЛНЕНИЕ ЗАДАЧ");
        developer.WriteCode("Новая система аутентификации");
        developer.CompleteTask("Разработка модуля пользователей");
        developer.CompleteProject("Веб-портал компании");
        
        designer.CreateDesign("Логотип новой продукции");
        designer.CompleteDesignProject("Ребрендинг компании");
        
        manager.ConductTeamMeeting();
        manager.ReviewPerformance();
        
        Console.WriteLine("\nОТПУСКА");
        developer.RequestVacation(DateTime.Now.AddDays(10), 14);
        designer.RequestVacation(DateTime.Now.AddDays(5), 7);
        
        Console.WriteLine("\nДОПОЛНИТЕЛЬНАЯ ИНФОРМАЦИЯ ");
        Console.WriteLine($"Стаж {manager.Name} в компании: {manager.CalculateYearsInCompany()} лет");
        Console.WriteLine($"Годовой бонус менеджера: {manager.CalculateBonus()} рублей");
        Console.WriteLine($"Портфолио дизайнера: {designer.PortfolioLink}");        
```
